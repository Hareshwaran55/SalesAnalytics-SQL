```python
%%sql    --deleting duplicates
delete from customers
where customer_id in(select customer_id from(
    select customer_id,
    row_number() over (partition by customer_name,gender,age,email,phone order by customer_id) as rn
    from customers
)t
    where rn>1
);
```

```python
%%sql
start transaction;
delete from products
where product_id in(select product_id from(
select product_id,
    row_number() over (partition by product_name,category,supplier,unit_price,stock_qty order by product_id) as rn
    from products
)t
                    where rn>1
);
```

```python
%%sql
start transaction;
delete from sales
where order_id in(select order_id from(
select order_id,
    row_number() over(partition by customer_id,product_id,region order by order_id) as rn
    from sales
)t
                  where rn>1
);
```

```python
%%sql --sales by country wise--
select c.country,
concat('$',format(sum(s.sales_amount),2)) as revenue,count(*) as order_numbers
from sales s
join customers c on s.customer_id=c.customer_id
group by c.country
order by sum(s.sales_amount) desc
limit 2;
```

```python
%%sql --quantity sold by product wise--
select p.product_name,count(s.quantity) as total_quantiy_sold
from sales s
join products p on s.product_id=p.product_id
group by p.product_name
order by total_quantiy_sold desc
limit 3;
```

```python
%%sql  --revenue by category--
select p.category, concat('$',format(sum(s.sales_amount),2)) as total_revenue,count(*) as num_orders
from sales s
join products p on s.product_id=p.product_id
group by p.category
order by sum(s.sales_amount) desc
limit 5;
```

```python
%%sql --revenue by segments--
select c.segment,concat('$',format(sum(s.sales_amount),2)) as revenue
from sales s
join customers c on s.customer_id=c.customer_id
group by c.segment
order by sum(s.sales_amount) desc;
```

```python
%%sql  --ranking products based on revenue--
with product as(
select p.product_name,round(sum(s.sales_amount),2) as product_revenue
from sales s
join products p on s.product_id=p.product_id
group by p.product_name
),
ranked_product as(
select product_name,product_revenue,
dense_rank() over (order by product_revenue desc) as product_rank
from product
)
select 
product_name,
concat('$', format(product_revenue,2)) as prodcut_revenue
from ranked_product
where product_rank=5;
```

```python
%%sql --customers by gender--
select c.gender,count(s.customer_id) as total_member
from sales s
join customers c on s.customer_id=c.customer_id
group by c.gender
order by total_member desc;
```

```python
%%sql --high value and low value customers--
with customers as(
select c.customer_name,round(sum(sales_amount),2) as sales_amount
from sales s
join customers c on s.customer_id=c.customer_id
group by c.customer_name
),
rank_customers as (
    select customer_name,concat('$',format(sales_amount,2)) as toatl_spend,
    dense_rank() over (order by sales_amount desc) as customer_rank
from customers
)
select * from rank_customers
where customer_rank<=5;
```

```python
%%sql --top selling produt by region--
with region as(
    select p.product_name,s.region,round(sum(sales_amount),2) as revenue,
dense_rank() over (partition by region order by sum(sales_amount) desc) as region_rank
from sales s
join products p on s.product_id=p.product_id
group by p.product_name,s.region
)
select region,product_name,concat('$',format(revenue,2)) as revenue
from region
where region_rank=1 and region != 'unknown';
```

```python
%%sql  --daily running sales--
with running_total as(
select
    str_to_date(order_date,'%d-%m-%Y') as order_date, round(sum(sales_amount),2) as daily_sales
from sales
group by order_date
)
select order_date, concat('$',format(daily_sales,2)) as daily_sales,
concat('$',format(sum(daily_sales) over(order by order_date),2)) as daily_revenue
from running_total
where year(order_date)='2022' and month(order_date)='01'
order by order_date
limit 2;
```

```python
%%sql  --montly revenue trend--
with monthly as(
select 
date_format(str_to_date(order_date,'%d-%m-%Y'),'%Y-%m') as month,
round(sum(sales_amount),2) as revenue
from sales
group by month
)
select month,
concat('$',format(revenue,2)) as revenue,
concat('$',format(revenue-lag(revenue) over(order by month),2)) as revenue_change
from monthly
order by month
limit 2;
```
